#### Act 3 Listas Con funciones para Influencer

```python
# ==========================================
# SISTEMA DE INFLUENCERS EN PYTHON
# ==========================================

influencers = []

# ==========================================
# FUNCIÓN AGREGAR
# ==========================================
def agregar_influencer():
    print("\n--- AGREGAR INFLUENCER ---")
    nombre = input("Nombre de usuario: ")
    plataforma = input("Plataforma: ")
    tematica = input("Temática: ")
    seguidores = input("Seguidores: ")

    inf = {
        "nombre": nombre,
        "plataforma": plataforma,
        "tematica": tematica,
        "seguidores": seguidores
    }

    influencers.append(inf)
    print(f"¡{nombre} ha sido agregado con éxito!")

# ==========================================
# FUNCIÓN LEER (POR NOMBRE)
# ==========================================
def leer_influencer():
    print("\n--- CONSULTAR INFLUENCER ---")
    if not influencers:
        print("La lista está vacía.")
        return

    buscar = input("Escriba el nombre del influencer que desea ver: ")
    encontrado = False

    for i in influencers:
        if i["nombre"].lower() == buscar.lower():
            print("\n--- DATOS ENCONTRADOS ---")
            print(f"Nombre: {i['nombre']}")
            print(f"Plataforma: {i['plataforma']}")
            print(f"Temática: {i['tematica']}")
            print(f"Seguidores: {i['seguidores']}")
            encontrado = True
            break # Salimos del bucle al encontrarlo

    if not encontrado:
        print(f"No se encontró a ningún influencer llamado '{buscar}'.")

# ==========================================
# FUNCIÓN LISTAR
# ==========================================
def listar_influencers():
    print("\n--- LISTA COMPLETA ---")
    if not influencers:
        print("No hay nadie en la lista todavía.")
    else:
        for i in influencers:
            print(f" >> {i['nombre']} ({i['plataforma']}) - {i['tematica']}")

# ==========================================
# FUNCIÓN MODIFICAR
# ==========================================
def modificar_influencer():
    print("\n--- MODIFICAR DATOS ---")
    buscar = input("Nombre del influencer a editar: ")

    for i in influencers:
        if i["nombre"].lower() == buscar.lower():
            print(f"Editando a {i['nombre']}. Deje en blanco para no cambiar.")

            nuevo_nombre = input(f"Nuevo nombre [{i['nombre']}]: ")
            nueva_plat = input(f"Nueva plataforma [{i['plataforma']}]: ")
            nueva_tema = input(f"Nueva temática [{i['tematica']}]: ")
            nuevos_seg = input(f"Nuevos seguidores [{i['seguidores']}]: ")

            if nuevo_nombre: i["nombre"] = nuevo_nombre
            if nueva_plat: i["plataforma"] = nueva_plat
            if nueva_tema: i["tematica"] = nueva_tema
            if nuevos_seg: i["seguidores"] = nuevos_seg

            print("Datos actualizados.")
            return

    print("Influencer no encontrado.")

# ==========================================
# FUNCIÓN BORRAR
# ==========================================
def borrar_influencer():
    print("\n--- ELIMINAR INFLUENCER ---")
    buscar = input("Nombre del influencer a borrar: ")

    for i in influencers:
        if i["nombre"].lower() == buscar.lower():
            influencers.remove(i)
            print(f"Se ha eliminado a {buscar} del sistema.")
            return

    print("No se pudo eliminar: Influencer no encontrado.")

# ==========================================
# MENÚ PRINCIPAL
# ==========================================
while True:
    print("\n" + "="*30)
    print("    MANAGER DE INFLUENCERS")
    print("="*30)
    print("1. Agregar nuevo")
    print("2. Ver datos de uno (por nombre)")
    print("3. Listar todos los nombres")
    print("4. Modificar datos")
    print("5. Eliminar")
    print("6. Salir")
    print("="*30)

    opcion = input("¿Qué quieres hacer?: ")

    if opcion == "1":
        agregar_influencer()
    elif opcion == "2":
        leer_influencer()
    elif opcion == "3":
        listar_influencers()
    elif opcion == "4":
        modificar_influencer()
    elif opcion == "5":
        borrar_influencer()
    elif opcion == "6":
        print("Apagando sistema...")
        break
    else:
        print("Esa opción no existe, intenta de nuevo.")


print("Programa Hecho por Arath Valenzuela NL52")
```
